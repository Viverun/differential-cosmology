# 2D Toy Forward Model\nGenerate initial conditions and evolve with the PM-lite solver.

```python
import jax.numpy as jnp\nimport matplotlib.pyplot as plt\n\nfrom diffcosmo.fields import generate_gaussian_field, power_law_power_spectrum\nfrom diffcosmo.pm import evolve_pm
```

```python
grid_shape = (64, 64)\ncosmology = {'Omega_m': 0.3, 'init_disp_scale': 0.25}\npower_fn = lambda k: power_law_power_spectrum(k, amplitude=1.0, index=-2.0)\n\ntheta0 = generate_gaussian_field(grid_shape, power_fn, seed=0)\ndelta_final = evolve_pm(theta0, n_steps=10, dt=0.1, cosmology=cosmology)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))\nax[0].imshow(theta0, cmap='RdBu_r')\nax[0].set_title('Initial field')\nax[1].imshow(delta_final, cmap='viridis')\nax[1].set_title('Evolved field')\nplt.tight_layout()
```
