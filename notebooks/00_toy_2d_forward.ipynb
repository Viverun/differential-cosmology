# 2D Forward Simulation

This notebook demonstrates generating initial conditions.

```python
import jax.numpy as jnp
import matplotlib.pyplot as plt
from diffcosmo import utils, fields
```

```python
# Generate k-space grid
grid_shape = (128, 128)
k = utils.compute_k_values(grid_shape)

print(f"Grid shape: {grid_shape}")
print(f"k range: [{k.min():.2f}, {k.max():.2f}]")
```

```python
P_k = fields.power_law_power_spectrum(k, amplitude=1.0, index=-2.0)

plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(jnp.log10(k + 1e-10), cmap='viridis')
plt.colorbar(label='log10(k)')
plt.title('Wavenumber k')

plt.subplot(1, 2, 2)
plt.imshow(jnp.log10(P_k + 1e-10), cmap='plasma')
plt.colorbar(label='log10(P(k))')
plt.title('Power Spectrum')
plt.tight_layout()
plt.show()
```
